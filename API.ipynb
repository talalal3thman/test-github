# API in Python

## What is an API?

An **API (Application Programming Interface)** is a set of rules and protocols that allows different software applications to communicate with each other. It acts as an intermediary that enables applications to send and receive data.

In **Python**, APIs are often used to:

- Fetch data from web services (e.g., weather data, stock prices, etc.).
- Interact with databases.
- Connect different software components.

**Top Common HTTP Status Codes**

| **Status Code** | **Meaning**                  | **Description** |
|---------------|------------------------------|---------------|
| **200**       | OK                            | Request was successful, and the server returned the expected response. |
| **201**       | Created                       | A new resource was successfully created (e.g., after a POST request). |
| **400**       | Bad Request                   | The request was invalid due to incorrect syntax or missing parameters. |
| **401**       | Unauthorized                  | Authentication is required to access the resource. |
| **403**       | Forbidden                     | Access to the resource is denied. |
| **404**       | Not Found                     | The requested resource does not exist on the server. |
| **500**       | Internal Server Error         | General server-side error. |
| **502**       | Bad Gateway                   | Received an invalid response from an upstream server. |
| **503**       | Service Unavailable           | Server is temporarily unavailable. |

```python
import requests
import pandas as pd

# Define the URL
url = "https://jsonplaceholder.typicode.com/posts"

try:
    # Sending GET request
    response = requests.get(url, timeout=10)

    # Raise an error for unsuccessful status codes
    response.raise_for_status()

    # Parse the JSON response
    data = response.json()

    if data:
        # Convert data directly to a DataFrame
        df = pd.json_normalize(data)
        print(df.head())
    else:
        print("No records found in the JSON response.")

except requests.exceptions.Timeout:
    print("Error: The request timed out.")

except requests.exceptions.ConnectionError:
    print("Error: Could not connect to the server.")

except requests.exceptions.HTTPError as e:
    print("HTTP Error:", e)

except requests.exceptions.RequestException as e:
    print("Request Error:", e)

except ValueError:
    print("Error: The response is not valid JSON.")

except Exception as e:
    print("Unexpected Error:", e)
```

```python
import requests
import pandas as pd

# Define the base URL for the API
url = "https://www.data.gov.bh/api/explore/v2.1/catalog/datasets/01-container-statistics-at-khalifa-bin-salman-port/records"

# Initialize parameters
params = {"limit": 100, "offset": 0}

# Empty list to store the data
all_data = []

# Loop 5 times to get 500 rows
for i in range(5):

    # Update offset
    params["offset"] = i * 100

    # Make the API request
    response = requests.get(url, params=params, timeout=100)

    # Convert response to JSON
    data = response.json()

    # Add results to the list
    all_data.extend(data["results"])

# Convert to DataFrame
df = pd.json_normalize(all_data)

df
```
