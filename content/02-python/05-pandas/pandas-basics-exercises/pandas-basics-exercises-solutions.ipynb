# Section III: Data Types and Dates

14) Create the DataFrame below as `sales`.

```python
{
    'invoice': ['A101', 'A102', 'A103', 'A104'],
    'quantity': ['5', '8', '3', '10'],
    'sale_date': ['2026-01-15', '2026-02-20', '2026-03-10', '2026-04-05']
}
```

Display the data type of every column.

````python
sales = pd.DataFrame({
    'invoice': ['A101', 'A102', 'A103', 'A104'],
    'quantity': ['5', '8', '3', '10'],
    'sale_date': ['2026-01-15', '2026-02-20', '2026-03-10', '2026-04-05']
})

sales.dtypes
````

15) Convert `quantity` from text to integer.

````python
sales['quantity'] = sales['quantity'].astype(int)
sales.dtypes
````

16) Convert `sale_date` to pandas datetime format.

````python
sales['sale_date'] = pd.to_datetime(sales['sale_date'])
sales.dtypes
````

17) Create a new column called `sale_year` by extracting the year from `sale_date`.

````python
sales['sale_year'] = sales['sale_date'].dt.year
sales
````

18) Create a new column called `sale_month` containing the full month name from `sale_date`.

````python
sales['sale_month'] = sales['sale_date'].dt.month_name()
sales
````

# Section IV: Creating, Renaming, Dropping and Unique Values

19) Create this DataFrame as `products`.

```python
{
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Mouse'],
    'category': ['Electronics', 'Accessories', 'Accessories', 'Electronics', 'Accessories'],
    'price': [450, 15, 25, 120, 18]
}
```

Create a new column named `stock` and assign the value `10` to every row.

````python
products = pd.DataFrame({
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Mouse'],
    'category': ['Electronics', 'Accessories', 'Accessories', 'Electronics', 'Accessories'],
    'price': [450, 15, 25, 120, 18]
})

products['stock'] = 10
products
````

20) Rename the `stock` column to `units_available`.

````python
products.rename(columns={'stock': 'units_available'}, inplace=True)
products
````

21) Delete the `units_available` column using `.drop()`.

````python
products.drop(columns=['units_available'], inplace=True)
products
````

22) From `products`, display the unique values in `category` and then return the number of unique categories.

````python
print(products['category'].unique())
print(products['category'].nunique())
````

23) Count how many times each value appears in the `category` column.

````python
products['category'].value_counts()
````

24) Calculate the percentage distribution of `category`, rounded to 2 decimal places.

````python
products['category'].value_counts(normalize=True).mul(100).round(2)
````

# Section V: Missing Values, Indexes and Duplicates

25) Create the DataFrame below as `customers`.

```python
{
    'customer_id': [1, 2, 3, 4, 4, 6],
    'name': ['Ali', 'Noor', 'Sara', 'Omar', 'Omar', 'Maya'],
    'age': [25, None, 31, 28, 28, None],
    'city': ['Manama', 'Riffa', None, 'Manama', 'Manama', 'Muharraq']
}
```

Count the null values in every column.

````python
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 4, 6],
    'name': ['Ali', 'Noor', 'Sara', 'Omar', 'Omar', 'Maya'],
    'age': [25, None, 31, 28, 28, None],
    'city': ['Manama', 'Riffa', None, 'Manama', 'Manama', 'Muharraq']
})

customers.isnull().sum()
````

26) Calculate the percentage of null values in every column, rounded to 2 decimal places.

````python
(customers.isnull().sum() / len(customers) * 100).round(2)
````

27) Return the index values of rows where `age` is null as a Python list.

````python
customers.index[customers['age'].isnull()].tolist()
````

28) Remove rows where `city` is null, then reset the index so it starts from `0`.

````python
customers_clean = customers.dropna(subset=['city']).reset_index(drop=True)
customers_clean
````

29) Fill the missing values in `age` using the median age.

````python
customers['age'] = customers['age'].fillna(customers['age'].median())
customers
````

30) Count duplicate values in `customer_id`, then remove duplicates based on `customer_id` while keeping the first occurrence.

````python
print("Duplicate customer_id values:", customers['customer_id'].duplicated().sum())

customers = customers.drop_duplicates(subset=['customer_id'], keep='first').reset_index(drop=True)
customers
````
