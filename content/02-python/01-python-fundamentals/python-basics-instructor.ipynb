## 3. String Operations


Strings in Python are versatile and come with many built-in methods.

### 3.1 Common String Methods

| Function                          | Description |
|-----------------------------------|-------------|
| `str.upper()`                     | Converts all characters to uppercase. |
| `str.lower()`                     | Converts all characters to lowercase. |
| `str.capitalize()`                 | Capitalizes the first letter of the string. |

| `str.strip(chars)`                 | Removes leading and trailing whitespace (or specified `chars`). |
| `str.lstrip(chars)`                | Removes leading whitespace (or specified `chars`). |
| `str.rstrip(chars)`                | Removes trailing whitespace (or specified `chars`). |
| `str.replace(old, new, count)`     | Replaces occurrences of `old` with `new` (optional: limit with `count`). |
| `str.find(sub, start, end)`        | Returns the index of the first occurrence of `sub`, or `-1` if not found. |
| `str.index(sub, start, end)`       | Returns the index of the first occurrence of `sub`, raises an error if not found. |
| `str.count(sub, start, end)`       | Counts occurrences of `sub` in the string. |
| `str.startswith(prefix, start, end)` | Returns `True` if the string starts with `prefix`. |
| `str.endswith(suffix, start, end)` | Returns `True` if the string ends with `suffix`. |
| `str.split(sep, maxsplit)`         | Splits the string into a list using `sep` (default: whitespace). |
| `str.join(iterable)`               | Joins elements of `iterable` into a string using `str` as a separator. |
| `str.isalpha()`                    | Returns `True` if all characters are letters. |
| `str.isdigit()`                    | Returns `True` if all characters are digits. |
| `str.isalnum()`                    | Returns `True` if all characters are letters or digits. |
| `str.isspace()`                    | Returns `True` if all characters are whitespace. |
| `str.swapcase()`                   | Swaps uppercase and lowercase characters. |
| `str.zfill(width)`                 | Pads the string with zeros on the left to match `width`. |

**8.2: While Loop (`while`)**
- Used when the number of iterations is not known beforehand.
- It will keep runs as long as the condition is `True`.


````python
count=0

while count<=5:
    print(count)
    count+=1
````

````python
number = 0
while True:
    number += 1
    print(number)
    if number == 10:  # Stop the loop when `number` equals 5
        break
````

````python
num = 0
while num <=5:
    num += 1
    if num == 3:
        continue  # Skip the rest of the loop when `num` is 3
    print(num)
````

````python
count = 10
while count >= 0:
    if count % 3 == 0:
        print(f"Stopping at {count}, which is divisible by 3.")
        break  # Exit the loop if the number is divisible by 3
    elif count % 2 == 0:
        print(f"{count} is even.")
    else:
        print(f"{count} is odd.")
    count -= 1  # Decrease count by 1 after each iteration
````

````python
# Number of rows for the triangle
num_rows = 5

# Initialize the current row
current_row = 1

# While loop to create the triangle
while current_row <= num_rows:
    # Print dots for the current row
    print('. ' * current_row)
    # Increment the current row
    current_row += 1
````

## 9. Functions in Python
- In Python, `def` is a **keyword** used to define a **function**.
- Functions are **reusable blocks of code** that perform a specific task.


````python
def sum_list(number_list):
    sum_nums = 0  
    for x in number_list:
        sum_nums += x  
    print(sum_nums)  
````
