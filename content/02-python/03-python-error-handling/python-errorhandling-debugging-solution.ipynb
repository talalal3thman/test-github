## 1. Introduction

## 2. Types of Python Errors

Python errors generally fall into two categories|

1. **Syntax Errors**: These occur when Python can't understand your code due to incorrect syntax (like forgetting a colon, mismatched parentheses, etc.). Python detects these before your code runs.

2. **Exceptions**: These occur during execution when something unexpected happens (like trying to divide by zero, accessing a non-existent file, etc.). Python provides many built-in exception types to help identify what went wrong.

## 3. Common Python Exceptions

Here are some of the most common exceptions you'll encounter:
|Error| Description |

| `IndexError`| Index out of range |

## 5. Handling Exceptions

### 5.1 ` try`/`except` Block

Python's `try`/`except` blocks allow you to handle potential errors gracefully:

```python
try:
    # code that might cause an exception
except SomeException:
    # what to do if that exception occurs
```

````python
# Example
try:
    age = int(input("Enter your age: "))
    print("Your age is", age)
except ValueError:
    print("Please enter a whole number.")
````

**What happened?**

Python first tries the code inside `try`.

- If everything works, the `except` block is skipped.
- If a `ValueError` occurs, Python moves to the matching `except` block.
- The program can then continue instead of stopping with an unhandled exception.

### 5.2 `try`/`except`/`else` Block

`else` runs only when the `try` block finishes **without an exception**.

```python
try:
    # code that might cause an exception
except SomeException:
    # what to do if that exception occurs
else:
    # what to do after if no exceptions
```

````python
# Example
try:
    age = int(input("Enter your age: "))
except ValueError:
    print("That is not a valid age.")
else:
    print(f"Your age in months is {age*12}")
    print(f"Your age in days is {age*365}")
````

Example of keep asking for valid input:

````python
while True:
    try:
        age = int(input("Enter your age: "))
    except ValueError:
        print("That is not a valid number. Try again.")
    else:
        break  # The loop breaks ONLY when a valid integer is entered successfully!

print(f"Your age is {age}.")
````

### 5.3 `raise` Block

So far, Python has raised exceptions for us.

We can also deliberately raise an exception using `raise`.

````python
try:
    age = int(input("Enter your age: "))
    if age < 0:
        raise ValueError("Age cannot be negative.")
except ValueError as e:
    print(f'Error: {e}')
else:
    print("Thank you. Your age is", age)
````

The advantage is that a function can reject invalid input immediately and clearly.

````python
def calculate_average(numbers):
    # Check for empty list to avoid division by zero
    if len(numbers)==0:
        return 0  # or return None, or any default value that makes sense
    try:
        total = sum(numbers)
        average = total / len(numbers)
        return average
    except TypeError:
        print("Error: All elements in the list must be numbers.")
        return None
````

### 6.4 Safe Division

````python
def safe_division(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print("Error: Division by zero is not allowed.")
        return None

# Test the function
print(safe_division(10, 2))  # Should return 5.0
print(safe_division(10, 0))  # Should handle error and return None
````

### 6.5 Age Group

````python
def classify_age(age_str):
    try:
        age = int(age_str)
    except ValueError:
        print(f"Invalid age: {age_str}")
        return None
    else:
        if age < 13:
            group = "child"
        elif age < 20:
            group = "teenager"
        else:
            group = "adult"
        print(f"Age {age} -> {group}")
        return group

for value in ["25", "7", "fifteen", "14"]:
    classify_age(value)
````

### 6.7 Validating Temperature

````python
def validate_temperature(temp):
    if temp < -50 or temp > 60:
        raise ValueError(f"Temperature {temp} is out of the valid range (-50 to 60).")
    return temp

temps = [25, -60, 40, 100, -10]

for t in temps:
    try:
        valid_temp = validate_temperature(t)
    except ValueError as e:
        print("Error:", e)
    else:
        print("Valid temperature:", valid_temp)
````

## 7. Advanced Error Handling

````python
def calculator(operation, a, b):
    """Calculate a result based on operation (add, subtract, multiply, divide).
    Handle all potential errors with appropriate messages."""
    try:
        # Check if a and b are numbers
        a_float = float(a)
        b_float = float(b)
        
        # Process based on operation
        if operation == 'add':
            return a_float + b_float
        elif operation == 'subtract':
            return a_float - b_float
        elif operation == 'multiply':
            return a_float * b_float
        elif operation == 'divide':
            return a_float / b_float
        else:
            raise ValueError(f"Unknown operation: {operation}")
    except TypeError:
        return f"Error: Both inputs must be numbers"
    except ValueError as e:
        if str(e).startswith("Unknown operation"):
            return f"Error: {e}"
        else:
            return f"Error: Could not convert inputs to numbers"
    except ZeroDivisionError:
        return f"Error: Cannot divide by zero"

# Test the function
print(calculator('add', 5, 3))       # Should return 8
print(calculator('divide', 10, 2))   # Should return 5.0
print(calculator('divide', 10, 0))   # Should handle division by zero
print(calculator('multiply', 4, 'x')) # Should handle type error
print(calculator('power', 2, 3))     # Should handle invalid operation
````
