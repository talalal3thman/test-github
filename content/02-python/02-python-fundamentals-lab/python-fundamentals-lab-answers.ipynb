108) Create a function `smallest_value(numbers)` that returns the smallest number in a list without using `min()`.


````python
def smallest_value(numbers):
    smallest = numbers[0]

    for number in numbers:
        if number < smallest:
            smallest = number

    return smallest

print(smallest_value([8, 3, 11, 2, 7]))
````

110) Create a function `count_vowels(text)` that counts and returns the number of vowels in a string without using `.count()`.


````python
def count_vowels(text):
    count = 0
    vowels = "aeiouAEIOU"

    for character in text:
        if character in vowels:
            count += 1

    return count

print(count_vowels("Python is useful"))
````
