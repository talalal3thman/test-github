## Brief:
In this exercise, you will simulate basic banking operations. Your program should allow a user to:
- Withdraw money.
- Deposit money.
- Check the current balance.
- View the mini-statement (transaction history).
- Exit the simulation.

Your account should start with a balance of $500. Each time a transaction is made, update the balance and save the transaction in a mini-statement. The user can continue making transactions until they choose/type "exit."

## Requirements:
Write a Python function to simulate basic bank operations:
- **Withdraw:** Ask the user for an amount to withdraw. Ensure the amount does not exceed the current balance.
- **Deposit:** Ask the user for an amount to deposit. Add it to the current balance.
- **Check Balance:** Display the current balance.
- **View Mini-statement:** Show the user all previous transactions (withdrawals and deposits).
- **Exit:** End the simulation when the user chooses/types "exit."

### Try-except block

```python
try:
    # User inputs two numbers
    numerator = int(input("Enter the numerator: "))
    denominator = int(input("Enter the denominator: "))
    
    # Perform the division
    result = numerator / denominator
    print(f"The result is: {result}")

except ZeroDivisionError:
    # Handle division by zero
    print("Error: You cannot divide by zero!")

except ValueError:
    # Handle invalid input (non-integer values)
    print("Error: Please enter valid integers.")

except Exception as e:
    # Handle any other exceptions
    print(f"An unexpected error occurred: {e}")
```
