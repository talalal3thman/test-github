# 1. Optimization Problem

## Example: Maximizing the Area of a Farm

A farmer has **100 meters of fencing** and wants to build a rectangular enclosure.

**Question:** What should the length and width be to maximize the enclosed area?

### Step 1: Define the Objective Function

We want to **maximize the area** of the rectangle:

$$
A = L \times W
$$

Where:
- $L$ = Length
- $W$ = Width

### Step 2: Define the Constraint

The farmer only has **100 meters of fencing**.

The perimeter of a rectangle is:

$$
2L + 2W = 100
$$

Rearranging to isolate $W$:

$$
W = 50 - L
$$

### Step 3: Rewrite the Objective Function

Substitute the constraint into the area formula:

$$
A(L) = L(50 - L)
$$

Simplifying:

$$
\boxed{A(L) = 50L - L^2}
$$

Now we have an objective function with only **one variable**.

**Goal:** Find the value of $L$ that maximizes the area, with $0 < L < 50$.

### Step 4: Find the Maximum Using Calculus

Take the derivative of the objective function:

$$
A'(L) = 50 - 2L
$$

At the maximum, the derivative equals zero:

$$
50 - 2L = 0
$$

Solving:

$$
L = 25
$$

Calculate the width:

$$
W = 50 - 25 = 25
$$

The second derivative is negative:

$$
A''(L) = -2
$$

This confirms that we found a maximum.

### Step 5: Calculate the Maximum Area

$$
A = 25 \times 25 = \boxed{625 \text{ m}^2}
$$

**Optimal solution:**
- Length = 25 meters
- Width = 25 meters
- Maximum area = 625 m²

### Connection to Gradient Descent

This example illustrates three important concepts:

1. **Objective Function:** The mathematical function we want to optimize.
2. **Constraint:** A restriction on the possible solutions.
3. **Optimization:** Finding the input that produces the best result.

We used calculus to find the exact maximum by setting the derivative to zero.

In Machine Learning, we usually want to **minimize a loss function**, rather than maximize an area.

**Gradient Descent** is an iterative optimization algorithm that updates model parameters to reduce the loss.

## 2.1 Newton's Method (Second-Order Optimization)

**Newton's Method:**

$$
x_{k+1} = x_k - \frac{1}{f''(x_k)}f'(x_k)
$$

### How Does It Work?

- **First derivative $f'(x)$:** Tells us the slope (direction and steepness).
- **Second derivative $f''(x)$:** Tells us the curvature (how quickly the slope changes).
- **High positive curvature:** The slope changes quickly, so Newton's Method takes smaller steps for the same gradient.
- **Low positive curvature:** The slope changes slowly, so Newton's Method takes larger steps for the same gradient.
- **Key idea:** Instead of manually choosing a fixed learning rate $\alpha$, Newton's Method uses $1/f''(x)$ to automatically scale the step based on curvature.
