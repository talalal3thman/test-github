## Overview

In the field of data science, good projects are practical. Your capstone project should be manageable and affect a real world audience. This might be a domain you are familiar with, a particular interest you have, something that affects a community you are involved in, or an area that relates to a field you wish to work in.

## Suggested Ways to Get Started
Begin by Asking:
- What is the scope of the need or problem I wish to investigate?
- Who is this for? Who is impacted or affected by this data? Who would benefit from this model? What are my goals for this investigation?
- What does success look like? How will I know if my model performs well?
- Where will I find data for this project? Is the data available?
- What format is the data in? What specific steps do I need to take to access it? How will I explain this project to outside audiences?

Other Tips:

- For your 1st potential topic, start with an idea, then look for potential data that could be used to support that idea.

- For your 2nd potential topic, reverse the process; look for interesting data and then extrapolate problems it could solve and audiences it could impact.

## Sample Scenario

### Problem Statement

A real-estate company needs a more consistent way to estimate the market value of residential properties. Current estimates may rely heavily on manual judgment and may vary between agents.  
Using historical property data such as location, size, number of bedrooms, bathrooms, age, and other property characteristics, we want to build a machine-learning model that predicts the expected selling price of a property.

_This is a strong Problem Statement because it explains:_

> the business problem → why it matters → what ML will do.


### Audience

- **Real-estate agents** — to support property valuations.
- **Property owners/sellers** — to understand an appropriate listing price.
- **Buyers** — to assess whether a property is fairly priced.
- **Business managers** — to improve consistency and efficiency of valuations.


### Goals
- Build a regression model that can estimate house prices from property characteristics.
- Explore which property features may be useful for predicting price.
- Understand the relationships between features and house prices.
- Identify suitable regression models to test later.
- Deploy model online...
